```python
import numpy as np
import pandas as pd
import os
import scanpy as sc
import anndata as ann
```

```python
PT1 = pd.read_csv('mtscATAC-seq/rds_output/clone_definition/PT1.tsv',sep='\t')
```

```python
PT1.rename(columns={'cell_id':'barcode'},inplace=True)
```

```python
PT1
```

```python
PT1_pos = sc.read_h5ad(os.path.join('mtscATAC-seq/h5_files/PT1_pos_pp.h5ad'))
```

```python
PT1_pos.obs
```

```python
PT1_pos.obs.reset_index(drop=False,inplace=True) 
PT1_pos.obs.rename(columns={'index':'barcode'}, inplace=True)
```

```python
PT1_pos.obs
```

```python
PT1_pos.obs = pd.merge(PT1_pos.obs,PT1,on='barcode',how='left')
```

```python
PT1_pos.obs
```

```python
PT2 = pd.read_csv('mtscATAC-seq/rds_output/clone_definition/PT2.tsv',sep='\t')
PT2.rename(columns={'cell_id':'barcode'},inplace=True)
```

```python
PT2_pos = sc.read_h5ad(os.path.join('mtscATAC-seq/h5_files/PT2_pos_pp.h5ad'))
```

```python
PT2_pos.obs.reset_index(drop=False,inplace=True) 
PT2_pos.obs.rename(columns={'index':'barcode'}, inplace=True)
```

```python
PT2_pos.obs = pd.merge(PT2_pos.obs,PT2,on='barcode',how='inner')
```

```python
PT1_pos_mg = sc.read_h5ad(os.path.join('mtscATAC-seq/h5ad_files/PT1_pos_mg.h5ad'))
PT2_pos_mg = sc.read_h5ad(os.path.join('mtscATAC-seq/h5ad_files/PT2_pos_mg.h5ad'))

PT1_pos_mg.obs = pd.merge(PT1_pos_mg.obs,PT1,on='barcode',how='inner')
PT2_pos_mg.obs = pd.merge(PT2_pos_mg.obs,PT2,on='barcode',how='inner')
```
